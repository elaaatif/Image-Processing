# TP1 of Image Processing

This notebook includes image reshaping, reading, and other operations using OpenCV.

```python
import cv2
import matplotlib.pyplot as plt
```

Question 1

```python

# Load the image from the file
image_path = 'rose_1024.tif'
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)  # Load in grayscale
```

```python

# Get the size of the image (width, height)
image_size = image.shape  # shape gives (height, width) in grayscale

# Get pixel values at specific coordinates (0,0), (10,10), and (15,30)
pixels = {
    (0, 0): image[0, 0],
    (10, 10): image[10, 10],
    (15, 30): image[15, 30]
}
```

```python

# Save the image in a different format (PNG) with a new filename
new_image_path = 'rose_1024_converted.png'
cv2.imwrite(new_image_path, image)
```

Question 2

```python
# Flip the image vertically
flipped_image = cv2.flip(image, 0)
```

```python

# Crop the region of the image defined by the rectangle [257:768, 257:768]
cropped_image = flipped_image[257:768, 257:768]
```

```python

# Resample the image by reducing its size by half
resized_image = cv2.resize(flipped_image, (image.shape[1] // 2, image.shape[0] // 2))
```

```python

# Extract intensity values of the middle horizontal line (512th row)
middle_line_profile = flipped_image[512, :]
```

```python
flipped_image_with_line = cv2.cvtColor(flipped_image, cv2.COLOR_GRAY2BGR)
cv2.line(flipped_image_with_line, (0, 512), (flipped_image.shape[1], 512), (255, 0, 0), 1)
plt.subplot(1, 3, 1)
plt.title("Flipped Image with 512th Line in Red")
plt.imshow(flipped_image_with_line)
plt.axis('off')
```

```python

# Display results
# Show flipped, cropped, and resized images
plt.figure(figsize=(12, 6))

plt.subplot(1, 3, 1)
plt.title("Vertically Flipped Image")
plt.imshow(flipped_image, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Cropped Region [257:768, 257:768]")
plt.imshow(cropped_image, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Resampled Image (Half Size)")
plt.imshow(resized_image, cmap='gray')
plt.axis('off')

# Display the middle horizontal line profile
plt.figure(figsize=(10, 4))
plt.plot(middle_line_profile, color='blue')
plt.title("512th Line Profile (Intensity Values)")
plt.xlabel("Position along the line")
plt.ylabel("Intensity")
plt.grid()
plt.show()
```
